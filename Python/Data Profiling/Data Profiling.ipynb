## Data Profiling

```python
import pandas as pd
import numpy as np
def data_profiling_report(df, table_name):

    print("="*100)
    print(f"DATA PROFILING REPORT : {table_name}")
    print("="*100)

    # ==========================================================
    # 1. DATASET OVERVIEW
    # ==========================================================

    print("\n1. DATASET OVERVIEW")
    print("-"*100)

    print(f"Rows               : {df.shape[0]}")
    print(f"Columns            : {df.shape[1]}")
    print(f"Memory Usage (KB)  : {round(df.memory_usage(deep=True).sum()/1024,2)}")


    # ==========================================================
    # 2. COLUMN INFORMATION
    # ==========================================================

    print("\n2. COLUMN INFORMATION")
    print("-"*100)

    column_info = pd.DataFrame({

        "Column Name":df.columns,
        "Data Type":df.dtypes.astype(str),
        "Non Null Count":df.count().values

    })

    print(column_info)


    # ==========================================================
    # 3. MISSING VALUES
    # ==========================================================

    print("\n3. MISSING VALUES")
    print("-"*100)

    missing = pd.DataFrame({

        "Missing Count":df.isnull().sum(),
        "Missing %":round(df.isnull().sum()/len(df)*100,2)

    })

    print(missing)


    # ==========================================================
    # 4. DUPLICATE ANALYSIS
    # ==========================================================

    print("\n4. DUPLICATE ANALYSIS")
    print("-"*100)

    print("Duplicate Rows :",df.duplicated().sum())
    print("Duplicate Primary Keys :",df.iloc[:,0].duplicated().sum())


    # ==========================================================
    # 5. UNIQUE VALUES
    # ==========================================================

    print("\n5. UNIQUE VALUES")
    print("-"*100)

    unique = pd.DataFrame({

        "Unique Values":df.nunique()

    })

    print(unique)


    # ==========================================================
    # 6. DESCRIPTIVE STATISTICS
    # ==========================================================

    print("\n6. DESCRIPTIVE STATISTICS")
    print("-"*100)

    print(df.describe(include="all"))


    # ==========================================================
    # 7. FREQUENCY DISTRIBUTION
    # ==========================================================

    print("\n7. FREQUENCY DISTRIBUTION")
    print("-"*100)

    object_cols = df.select_dtypes(include=["object","category"]).columns

    if len(object_cols)==0:

        print("No Categorical Columns Found")

    else:

        for col in object_cols:

            print(f"\nColumn : {col}")

            if df[col].nunique()<=20:

                print(df[col].value_counts(dropna=False))

            else:

                print(f"Too many unique values ({df[col].nunique()}). Top 10 shown below.")
                print(df[col].value_counts(dropna=False).head(10))


    # ==========================================================
    # 8. DATE ANALYSIS
    # ==========================================================

    print("\n8. DATE ANALYSIS")
    print("-"*100)

    date_cols = df.select_dtypes(include=["datetime64[ns]"]).columns

    if len(date_cols)==0:

        print("No Date Columns Found")

    else:

        for col in date_cols:

            print(f"\nColumn : {col}")
            print("Minimum Date :",df[col].min())
            print("Maximum Date :",df[col].max())
            print("Missing Dates :",df[col].isnull().sum())


    # ==========================================================
    # 9. OUTLIER DETECTION
    # ==========================================================

    print("\n9. OUTLIER DETECTION")
    print("-"*100)

    numeric_cols = df.select_dtypes(include=np.number).columns

    if len(numeric_cols)==0:

        print("No Numeric Columns Found")

    else:

        for col in numeric_cols:

            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)

            IQR = Q3-Q1

            Lower = Q1-(1.5*IQR)
            Upper = Q3+(1.5*IQR)

            Outliers=((df[col]<Lower)|(df[col]>Upper)).sum()

            print(f"{col} : {Outliers} Outliers")


    # ==========================================================
    # 10. INVALID VALUES
    # ==========================================================

    print("\n10. INVALID VALUES")
    print("-"*100)

    print("Blank Strings in Text Columns")

    for col in object_cols:

        blanks = df[col].astype(str).str.strip().eq("").sum()

        print(f"{col} : {blanks}")


    # ==========================================================
    # 11. DATA CONSISTENCY
    # ==========================================================

    print("\n11. DATA CONSISTENCY")
    print("-"*100)

    for col in object_cols:

        original = df[col].nunique(dropna=True)

        standardized = (
            df[col]
            .astype(str)
            .str.strip()
            .str.title()
            .nunique(dropna=True)
        )

        print(f"{col}")
        print(f"Original Unique : {original}")
        print(f"After Standardization : {standardized}\n")


    # ==========================================================
    # 12. TEXT QUALITY
    # ==========================================================

    print("\n12. TEXT QUALITY")
    print("-"*100)

    for col in object_cols:

        leading = df[col].astype(str).str.startswith(" ").sum()

        trailing = df[col].astype(str).str.endswith(" ").sum()

        blank = df[col].astype(str).str.strip().eq("").sum()

        print(f"\nColumn : {col}")
        print("Leading Spaces :",leading)
        print("Trailing Spaces :",trailing)
        print("Blank Values :",blank)


    # ==========================================================
    # 13. DATA COMPLETENESS
    # ==========================================================

    print("\n13. DATA COMPLETENESS")
    print("-"*100)

    completeness = pd.DataFrame({

        "Complete %":round(df.notnull().sum()/len(df)*100,2)

    })

    print(completeness)

    overall = round((df.notnull().sum().sum()/(df.shape[0]*df.shape[1]))*100,2)

    print(f"\nOverall Dataset Completeness : {overall}%")

    print("\n")
    print("="*100)
    print("END OF REPORT")
    print("="*100)
```

```python
def data_quality_summary(df, table_name):

    object_cols = df.select_dtypes(include=["object", "category"]).columns
    numeric_cols = df.select_dtypes(include="number").columns

    # Convert date columns automatically
    for col in df.columns:
        if "date" in col.lower():
            df[col] = pd.to_datetime(df[col], errors="coerce")

    date_cols = df.select_dtypes(include=["datetime64[ns]"]).columns

    rows = df.shape[0]
    columns = df.shape[1]

    missing_values = df.isnull().sum().sum()
    duplicate_rows = df.duplicated().sum()
    duplicate_pk = df.iloc[:,0].duplicated().sum()

    blank_strings = 0
    for col in object_cols:
        blank_strings += df[col].astype(str).str.strip().eq("").sum()

    future_dates = 0
    for col in date_cols:
        future_dates += (df[col] > pd.Timestamp.today()).sum()

    outliers = 0

    for col in numeric_cols:

        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)

        iqr = q3-q1

        lower = q1-(1.5*iqr)
        upper = q3+(1.5*iqr)

        outliers += ((df[col]<lower)|(df[col]>upper)).sum()

    completeness = round(
        (df.notnull().sum().sum()/(rows*columns))*100,
        2
    )

    print(f"{table_name:<28} {rows:>7} {columns:>5} {missing_values:>8} {duplicate_rows:>8} {duplicate_pk:>8} {outliers:>10} {completeness:>10}%")
```

```python
print("="*110)
print(f"{'TABLE NAME':<28} {'ROWS':>7} {'COLS':>5} {'MISSING':>8} {'DUP_ROWS':>8} {'DUP_PK':>8} {'OUTLIERS':>10} {'COMPLETE':>10}")
print("="*110)

for table_name, df in tables.items():
    data_quality_summary(df, table_name)

print("="*110)
```
